# deepEmulator — local visible inference

This notebook is **local-only** (PyBoy SDL2 + pygame don't render in Colab). Use it as a recipe for pulling a Drive-trained bundle and watching the agent play with optional arrow + attention overlays.

**Prereqs:**
- A trained bundle synced from Drive (`rclone copy gdrive:deepEmulator/checkpoints/pokemon_red/<run> ./checkpoints/pokemon_red/<run>`)
- `pip install -e ".[viz]"` for pygame
- `roms/PokemonRed.gb` + `states/init.state` locally

## Quick-start (shell)

```bash
rclone copy gdrive:deepEmulator/checkpoints/pokemon_red ./checkpoints/pokemon_red
RUN=$(cat ./checkpoints/pokemon_red/latest.txt)

deepemu-play \
    --cartridge "POKEMON RED" \
    --rom roms/PokemonRed.gb \
    --init-state states/init.state \
    --ckpt $RUN \
    --visible --arrows --attention
```

Three windows open:
1. The PyBoy SDL2 window — the game itself, playing in real time.
2. The **arrows** window — direction-colored trajectory accumulating as the agent moves.
3. The **attention** window (if the bundle has a DINO encoder) — alpha-blended attention heatmap overlay on the current frame.

## Human takeover

While the agent is playing, write `0` (or empty) to `<RUN>/agent_enabled.txt` to pause the agent (sends NOOP every step; you can play with the keyboard via the PyBoy window). Write `1` to resume.

```bash
echo 0 > $RUN/agent_enabled.txt   # human takeover
echo 1 > $RUN/agent_enabled.txt   # back to agent
```

## Programmatic use

If you want to embed in a script:

````python
from pathlib import Path
from deepEmulator.inference.play import run_play

run_play(
    cartridge='POKEMON RED',
    rom=Path('roms/PokemonRed.gb'),
    init_state=Path('states/init.state'),
    ckpt=Path('checkpoints/pokemon_red'),  # parent dir — resolves via latest.txt
    visible=True,
    arrows=True,
    attention=True,
    episodes=3,
    max_episode_steps=2048,
    epsilon=0.05,
)
````
