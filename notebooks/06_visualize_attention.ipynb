# deepEmulator — Attention rollout visualization

Renders an animated GIF of the DINO encoder's CLS attention overlaid on a Pokemon Red episode. Uses multi-head attention rollout (Abnar & Zuidema 2020) — per-layer head-averaged attention is combined with the identity (residual connection) and chained layer-by-layer to produce CLS-to-patch attention.

Pairs with the trajectory-arrow visualization (notebook 03 / P5). Together they form the demo loop: watch the agent play + see what it's looking at (attention) + see where it's been (arrows).

```python
from pathlib import Path
from google.colab import drive; drive.mount('/content/drive', force_remount=False)
DRIVE = Path('/content/drive/MyDrive/deepEmulator')
encoder_dir = Path((DRIVE / 'encoders/dino/latest.txt').read_text().strip())
rom = DRIVE / 'roms/PokemonRed.gb'
init_state = DRIVE / 'states/init.state'
out_gif = DRIVE / 'attention_pokemon_red.gif'
print('encoder:', encoder_dir)
print('out:', out_gif)
```

```python
from deepEmulator.visualization.attention import main as attention_main
attention_main([
    '--encoder', str(encoder_dir),
    '--cartridge', 'POKEMON RED',
    '--rom', str(rom),
    '--init-state', str(init_state),
    '--steps', '300',
    '--alpha', '0.5',
    '--fps', '15',
    '--out', str(out_gif),
])
```
