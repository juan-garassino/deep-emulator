# deepEmulator — Trajectory arrow visualization

Renders an arrow-flow PNG over the global game map from a bundle's per-episode trajectory CSVs. Port of PWhiddy's `BetterMapVis_script_version_FLOW.py` — each visited tile gets an arrow rotated by the resultant movement vector, colored by direction (HSL hue ramp). Pairs with `notebook 06` (attention overlay): arrows = where the agent went, attention = what the agent was looking at.
